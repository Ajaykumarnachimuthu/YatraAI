# YĀTRĀ AI — Notebook 02: Exploratory Data Analysis (EDA)
## Multi-Modal Network Topology, Fare Curves & Transit Delay Patterns

This notebook explores the multimodal transit supply network across India:
- Railway network coverage and geographic distribution
- Train service categorization and halt structures
- Flight pricing distributions across trunk corridors
- Operational train delay distributions
- Monsoon rainfall seasonal risk patterns


```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

BASE_DIR = os.path.dirname(os.getcwd())
if BASE_DIR not in sys.path:
    sys.path.append(BASE_DIR)

PROCESSED_DIR = os.path.join(BASE_DIR, "data", "processed")
print("Libraries imported successfully.")
```

### 1. Railway Network: Station Geography & Service Density


```python
df_stations = pd.read_parquet(os.path.join(PROCESSED_DIR, "canonical_stations.parquet"))
print(f"Total Stations: {len(df_stations):,}")
print(f"Stations with Valid Indian Coordinates: {(df_stations['is_valid_coordinates']).sum():,}")
print(f"Flagged Dummy / Test Station Codes: {(df_stations['is_dummy_station']).sum():,}")

df_stations.head()
```

```python
plt.figure(figsize=(10, 4))
zone_counts = df_stations['railway_zone'].value_counts()
sns.barplot(x=zone_counts.index, y=zone_counts.values, palette='viridis')
plt.title("Indian Railway Stations by Railway Zone", fontsize=12, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.ylabel("Number of Stations")
plt.tight_layout()
plt.show()
```

### 2. Commercial Aviation: Domestic Fare Dynamics Across Top 6 Metros


```python
df_flights = pd.read_parquet(os.path.join(PROCESSED_DIR, "canonical_flights.parquet"))
print(f"Total Flight Records: {len(df_flights):,}")
print(f"Metro Corridors Covered: {df_flights['source_city'].nunique()} origins x {df_flights['destination_city'].nunique()} destinations")
df_flights[['airline', 'flight_number', 'source_city', 'destination_city', 'departure_time_window', 'price_inr', 'duration_hours']].head()
```

```python
plt.figure(figsize=(10, 4))
sns.boxplot(data=df_flights, x='airline', y='price_inr', hue='airline', palette='tab10', legend=False)
plt.title("Domestic Flight Price Distribution by Airline (INR)", fontsize=12, fontweight='bold')
plt.xticks(rotation=30, ha='right')
plt.ylim(0, 35000)
plt.ylabel("Price (INR)")
plt.tight_layout()
plt.show()
```

### 3. Historical Train Delay Risk Distributions


```python
df_delays = pd.read_parquet(os.path.join(PROCESSED_DIR, "canonical_train_delays.parquet"))
print(f"Total Halt Delay Observations: {len(df_delays):,}")
df_delays[['train_number', 'train_name', 'station_code', 'scheduled_arrival', 'avg_delay_arrival_min', 'p_on_time', 'p_severe_delay']].head()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df_delays['avg_delay_arrival_min'], bins=30, kde=True, color='crimson')
plt.title("Distribution of Average Arrival Delay at Corridor Stations (Minutes)", fontsize=12, fontweight='bold')
plt.xlabel("Average Arrival Delay (Minutes)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
